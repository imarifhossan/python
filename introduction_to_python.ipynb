Even, Odd, Positive and Negative

```python
n = int(input())
inp = input()
numbers = inp.split()


even = 0; odd = 0; positive = 0; negative = 0; 
for i in range(n):
    x = int(numbers[i])

    #positive or negative
    if(x>0):
          positive+=1
    elif(x<0):
          negative+=1

    #even or odd
    if(x%2==0):
         even+=1
    else:
         odd+=1
print("Even:",even) 
print("Odd:",odd) 
print("Positive:",positive) 
print("Negative:",negative) 
```

```python
#Mini Loss Tracker
n = int(input())
target = float(input())
sum = 0
inp_arr = []
for i in range(n):
    val = float(input())
    sum += val 
avg = sum / n
if (avg <= target):
    print("PASS")
else:
    print("RETRY")
```

```python
#Majority Vote Gate
n = int(input())
YES_count=0 
NO_count =0

for i in range(n):
    vote = input()
    if vote =="YES":
        YES_count+=1
    elif vote =="NO":
        NO_count+=1
if (YES_count >= NO_count):
    print("ACCEPT")
else:
    print("REJECT")

```

```python
#AI Brightness Analyzer
inp = input()
image_light= inp.split()
length = len(image_light)
sum = 0
for i in range(length):
    sum+=int(image_light[i])
avg = sum / length
if(avg<85):
    print("Dark Image")
elif(avg>=85 and avg<=170):
    print("Normal Image")
elif(avg>170):
    print("Bright Image")
```

```python
#AI Bias Detector
inp = input()
models = inp.split()
count_a=0;count_b=0
length = len(models)

for i in range(length):
    if(models[i]=='A'):
        count_a+=1
    elif(models[i]=='B'):
        count_b+=1
count_a= int(count_a*100/length)
count_b= int(count_b*100/length)

if (count_a>=70 or count_b>=70):
    print("Biased Model")
else:
    print("Fair Model")

```
