```python
#Declaration
A = {10,20,30, "John"}
print(A, type(A))
```

```python
#Empty declaration

a = []
print(type(a))

b = ()
print(type(b))

c = {}
print(type(c))

s = set()
print(type(s))
```

```python
#accessing (use in and for)

s= {1,2,3,4,5}

if (3 in s):
    print("3 is present")
else:
    print("3 is not present")

#iteration
sum=0
for element in s:
    sum+=element
print(sum)
```

```python
#add item
s.add(10)
s.add(0)
print(s)
#remove a random element from first
removedV=s.pop()
print(s)
print(removedV)

#remove specific element
s.remove(300)
print(s)

```

```python
#important methods of set
#union, intersection
#disjoints (nothing common) joint (something common)
#subset (when all the elements found in another set)


p = {1,2,3}
q = {3,4,5,6}
t= {1,2,3,4,5,6}
print(p.union(q))

print(p.intersection(q))
p.intersection_update(q)
print(p)
print(p.isdisjoint(q))
print(p.issubset(t))

#symmetric_difference
z = q ^ t
print(z)


#update
x = {'a','b','c'}
y = {'d','b'}
# x.update(y)
x.difference_update(y)
print(x)


```
