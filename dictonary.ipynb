Key Value pair
Mutable(Change possible)
Unordered

Key is immutable(int,string,float,tuple)
Value (can be anything, string, list, set, tuple)

```python
#declaration
dic = {}

print(type(dic))

dic = {'name': "Phitron",
"age":20,"address":"Dhaka","numbers":[10,20,30,40]}

print(dic)

#accessing
print(dic['name'])
print(dic.get('age'))

#updating
dic['name'] = 'AI and ML'
print(dic['name'])
#if the key is not available than by default it returns none 
print(dic.get('math_marks'))
print(dic.get('math_marks',0))#set the value 0

#But for this case it will return key error
#print(dic["math_marks"])


#age will update as 30 when its find new value 
#dic = {'name': "Phitron","age":20,"address":"Dhaka","numbers":[10,20,30,40],"age":30}
#print(dic)



```

```python
print(dic)
#adding new value
dic["math_marks"]=30
print(dic)

dic.update({"english_marks":40})
print(dic)

#deleting the key value
del dic["math_marks"]
print(dic)


dic2 = dic.copy()
print(dic2)
```
