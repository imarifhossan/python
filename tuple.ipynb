```python
#built in data type , where we can store different types of data , immutable, not reassign-able
tup = (10,20,30,40)
print(type(tup),tup)
```

```python
tup1 = (10.20,32.54,40.56)
print(type(tup1),tup1)
```

```python
tup3=("Apple","Banana")
print(type(tup3),tup3)
```

```python
tup4=(10,20,11.25,524,45, True, "Apple", "Pineapple")
print(type(tup4),tup4)
```

```python
tup5 = (10)
tup6 = (10,)
print(type(tup5),tup5)
print(type(tup6),tup6)
```

```python
numbers = [10,20,25.5]
tup7 = tuple(numbers)
print(tup7,type(tup7))
```

```python
#access
print(tup4[3])

#slicing
new_tup = tup7[0:2]
print(new_tup)
```
